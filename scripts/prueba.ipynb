```python
import cv2

ruta = "../Data/703 (2026-10-19 11'59'13 - 2026-10-19 12'02'13).avi"
cap = cv2.VideoCapture(ruta)

# 1. Verificar si abre
print("¿Abrió?", cap.isOpened())

# 2. Ver el códec (FOURCC). Si devuelve 0.0, es una mala señal.
fourcc = cap.get(cv2.CAP_PROP_FOURCC)
print("FOURCC (código de códec):", fourcc)

# 3. Intentar leer un frame real
ret, frame = cap.read()
print("¿Leyó un frame?", ret)
if ret:
    print("Dimensiones del frame:", frame.shape)

cap.release()
```
